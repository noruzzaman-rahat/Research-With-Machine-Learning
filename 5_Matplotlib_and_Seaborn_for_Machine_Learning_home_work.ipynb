```python
import pandas as pd

gym = pd.read_csv("/content/drive/MyDrive/skill morph/gym_membership.csv")
print(gym.head())
```

```python
#Matplotib
```

```python
import matplotlib.pyplot as plt

gym['abonoment_type'].value_counts().plot(kind='bar')
plt.title("Membership Types")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
```

```python
plt.hist(gym['avg_time_in_gym'], bins=10)
plt.title("Average Time in Gym")
plt.xlabel("Hours")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(gym['Age'], gym['visit_per_week'])
plt.title("Age vs Visit Per Week")
plt.xlabel("Age")
plt.ylabel("Visit Per Week")
plt.show()
```

```python
#Seaborn
```

```python
import seaborn as sns

sns.boxplot(x='gender', y='Age', data=gym)
plt.title("Gender vs Age")
plt.show()
```

```python
sns.countplot(x='days_per_week', data=gym)
plt.title("Days Per Week")
plt.show()
```

```python
sns.barplot(x='abonoment_type', y='visit_per_week', data=gym)
plt.title("Average Visit Per Week by Membership Type")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv(f'{DATA_DIR}/train (1).csv')

fig, axes = plt.subplots(2, 2, figsize=(11, 7))

# Histogram: Rating distribution
axes[0, 0].hist(df['rating'], bins=15)
axes[0, 0].set_title('Rating Distribution')

#Bar chart: Class count (top categories)
df['class_name'].value_counts().head(10).plot(kind='bar', ax=axes[0, 1])
axes[0, 1].set_title('Top Class Count')

# Scatter plot: Latitude vs Longitude (sample)
axes[1, 0].scatter(df['latitude'].head(100),
                   df['longitude'].head(100), alpha=0.6)
axes[1, 0].set(title='Latitude vs Longitude (sample)',
               xlabel='Latitude', ylabel='Longitude')

# Line plot: Rating (first 50 samples)
axes[1, 1].plot(df['rating'].head(50))
axes[1, 1].set_title('Rating — First 50 Samples')

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
sns.histplot(df['rating'], bins=20, kde=True, ax=ax)
ax.set_title('Rating — Histogram + KDE')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

#  Rating by class_name
top_classes = df['class_name'].value_counts().head(8).index
sns.boxplot(data=df[df['class_name'].isin(top_classes)],
            x='class_name', y='rating', ax=axes[0])
axes[0].set_title('Rating by Class Name')
axes[0].tick_params(axis='x', rotation=45)

# Rating by primary_label
top_labels = df['primary_label'].value_counts().head(8).index
sns.boxplot(data=df[df['primary_label'].isin(top_labels)],
            x='primary_label', y='rating', ax=axes[1])
axes[1].set_title('Rating by Primary Label')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

top_classes = df['class_name'].value_counts().head(10).index
sns.countplot(data=df[df['class_name'].isin(top_classes)],
              x='class_name', ax=axes[0])
axes[0].set_title('Class Name Count')
axes[0].tick_params(axis='x', rotation=45)


top_labels = df['primary_label'].value_counts().head(10).index
sns.countplot(data=df[df['primary_label'].isin(top_labels)],
              x='primary_label', ax=axes[1])
axes[1].set_title('Primary Label Count')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 4))

order = df.groupby('class_name')['rating'].mean().sort_values().index

sns.barplot(data=df[df['class_name'].isin(order[:10])],
            x='class_name', y='rating',
            order=order[:10], ax=ax)

ax.set_title('Average Rating by Class Name')
ax.tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots()

top_classes = df['class_name'].value_counts().head(5).index

sns.kdeplot(data=df[df['class_name'].isin(top_classes)],
            x='rating', hue='class_name',
            fill=True, ax=ax)

ax.set_title('Rating Density by Class Name')

plt.show()
```

```python
corr = df[['rating', 'latitude', 'longitude']].corr()

fig, ax = plt.subplots(figsize=(5, 4))

sns.heatmap(corr, annot=True, vmin=-1, vmax=1, fmt='.2f', ax=ax)

ax.set_title('Correlation Matrix')

plt.show()
```

```python
sns.pairplot(
    df[df['class_name'].isin(top_classes)][['rating', 'latitude', 'longitude', 'class_name']],
    hue='class_name',
    height=2.2
)

plt.show()
```

```python
def quick_eda(df):
    """Six-panel EDA summary for your dataset."""

    fig, axes = plt.subplots(2, 3, figsize=(13, 7))

    # 1️ Histogram: Rating
    axes[0, 0].hist(df['rating'].dropna(), bins=20)
    axes[0, 0].set_title('Rating Distribution')

    # 2️ Bar: Top class_name
    df['class_name'].value_counts().head(8).plot(
        kind='bar', ax=axes[0, 1])
    axes[0, 1].set_title('Top Class Names')
    axes[0, 1].tick_params(axis='x', rotation=45)

    # 3️ Pie: Primary label distribution
    labels = df['primary_label'].value_counts().head(5)
    axes[0, 2].pie(labels.values, labels=labels.index, autopct='%1.0f%%')
    axes[0, 2].set_title('Top Primary Labels')

    # 4️ Scatter: Latitude vs Longitude
    axes[1, 0].scatter(df['latitude'], df['longitude'], alpha=0.3)
    axes[1, 0].set(title='Location Scatter', xlabel='Latitude', ylabel='Longitude')

    # 5️ Bar: Avg rating by class_name
    top_classes = df['class_name'].value_counts().head(6).index
    df[df['class_name'].isin(top_classes)].groupby('class_name')['rating'].mean().plot(
        kind='bar', ax=axes[1, 1])
    axes[1, 1].set_title('Avg Rating by Class')
    axes[1, 1].tick_params(axis='x', rotation=45)

    # 6️ Mean of numerical columns
    df.select_dtypes(include='number').mean().plot(
        kind='bar', ax=axes[1, 2])
    axes[1, 2].set_title('Average Numeric Values')

    plt.tight_layout()
    plt.show()


# Run
df = pd.read_csv(f'{DATA_DIR}/train (1).csv')
quick_eda(df)
```
